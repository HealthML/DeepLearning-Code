## Gradient Descent

### Steps of the Gradient Descent Algorithm:

**Step 1:** Choose an initial value$x$ and a stepsize$\eta > 0$

**Step 2:** Compute gradient of the objective function

**Step 3:** Update$x$ using the following rule:

$$x_{i+1} = x_i - \eta \nabla f(x_i)$$

Repeat Steps 2 and Steps 3 until the stopping criterion is hold, for example$\nabla f$ is small enough or the number of iterations has reached a certain value.

## Why Gradient Descent algorithm converge to minimum of the objective function:

Assume that $f$ is a continuously differentiable real-valued function

$$f: \mathbb{R} \rightarrow \mathbb{R}$$

Using a Taylor expansion we obtain:

$$f(x + h) = f(x) + h f'(x) + O(h^2)$$

It is the first-order approximation of function$f$

Now we see that how moving to the direction of the $-\nabla f$ will decrease$f$

We keep a fixed step size $\eta > 0$ and set $h = -\eta f'(x)$. Plugging it into the Taylor expansion we get

$$f(x - \eta f'(x)) = f(x) - \eta f'(x)^2 + O(\eta^2 f'(x)^2)$$

Now if $f'(x) \neq 0$, we see that

$$f(x - \eta f'(x)) < f(x)$$

since $\eta f'(x)^2 > 0$

This means that if we use the following formula to update x:

$$x \leftarrow x - \eta f'(x)$$

the value of function $f(x)$ might decline.

##  Newton's Method 

Motivation: As you could see in GD, getting the learning rate $\eta$ "just right" is tricky.

- If we pick it too small, we make little progress.
- If we pick it too large, the solution oscillates and in the worst case it might even diverge.

What if we could determine $\eta$ automatically or get rid of having to select a learning rate at all?

Second-order methods that look not only at the value and gradient of the objective function but also at its curvature can help in this case.

Note that these methods cannot be applied to deep learning directly due to the computational cost, but they provide useful intuition into how to design advanced optimization algorithms that mimic many of the desirable properties of second-order methods.
